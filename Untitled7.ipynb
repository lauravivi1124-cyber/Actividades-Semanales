```python

# ANALISIS FINANCIERO DE TRES ACTIVOS CON YFINANCE

# Activos: XOM, WMT y ADBE

# 1. INSTALAR E IMPORTAR LIBRERIAS
!pip install -q yfinance pandas numpy matplotlib seaborn

import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuracion de las graficas
sns.set_theme(style="whitegrid")


# PUNTO 1. DESCARGA DE LOS TRES ACTIVOS CON YFINANCE

tickers = ["XOM", "WMT", "ADBE"]

# Descargar precios historicos de los ultimos 5 años
datos = yf.download(
    tickers,
    period="5y",
    interval="1d",
    auto_adjust=True,
    progress=False
)

# Seleccionar precios de cierre ajustados
precios = datos["Close"].copy()

# Conservar fechas con datos completos para los tres activos
precios = precios.dropna()

# Verificar que se descargaron datos
if precios.empty:
    raise ValueError(
        "No se descargaron precios. Verifica tu conexion "
        "a internet e intenta ejecutar nuevamente."
    )

print("=" * 60)
print("PUNTO 1: DESCARGA DE DATOS")
print("=" * 60)

print("\nPrimeras cinco observaciones:")
display(precios.head())

print("\nUltimas cinco observaciones:")
display(precios.tail())

print("\nCantidad de observaciones:", len(precios))
print("\nActivos descargados:", list(precios.columns))


# PUNTO 2. PRECIOS DE CIERRE Y NORMALIZACION BASE 100

print("\n" + "=" * 60)
print("PUNTO 2: PRECIOS DE CIERRE Y BASE 100")
print("=" * 60)

# Grafica de precios de cierre
plt.figure(figsize=(12, 6))

for ticker in tickers:
    plt.plot(
        precios.index,
        precios[ticker],
        label=ticker
    )

plt.title("Precios de cierre ajustados de XOM, WMT y ADBE")
plt.xlabel("Fecha")
plt.ylabel("Precio en dolares (USD)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Normalizar los precios tomando el primer dia como base 100
precios_base100 = precios.div(precios.iloc[0]).mul(100)

# Grafica de precios normalizados
plt.figure(figsize=(12, 6))

for ticker in tickers:
    plt.plot(
        precios_base100.index,
        precios_base100[ticker],
        label=ticker
    )

plt.axhline(
    y=100,
    color="black",
    linestyle="--",
    label="Base 100"
)

plt.title("Evolucion de los activos: indice base 100")
plt.xlabel("Fecha")
plt.ylabel("Indice base 100")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Comparar valores iniciales y finales
print("\nValores normalizados al inicio y al final:")
display(precios_base100.iloc[[0, -1]].round(2))

# Calcular rentabilidad acumulada porcentual
rentabilidad_acumulada = (
    (precios.iloc[-1] / precios.iloc[0]) - 1
) * 100

print("\nRentabilidad acumulada del periodo (%):")
display(rentabilidad_acumulada.round(2).to_frame(
    name="Rentabilidad acumulada (%)"
))

# Interpretacion automatica
mejor_activo = rentabilidad_acumulada.idxmax()
peor_activo = rentabilidad_acumulada.idxmin()

print(
    f"\nEl activo con mayor rentabilidad acumulada fue "
    f"{mejor_activo}, con "
    f"{rentabilidad_acumulada[mejor_activo]:.2f}%."
)

print(
    f"El activo con menor rentabilidad acumulada fue "
    f"{peor_activo}, con "
    f"{rentabilidad_acumulada[peor_activo]:.2f}%."
)

print(
    "\nInterpretacion: el indice base 100 permite comparar "
    "el crecimiento porcentual de los activos desde una "
    "misma base inicial."
)


# PUNTO 3. RENDIMIENTOS DIARIOS, HISTOGRAMAS Y BOXPLOT

print("\n" + "=" * 60)
print("PUNTO 3: RENDIMIENTOS DIARIOS")
print("=" * 60)

# Calcular rendimientos diarios porcentuales
rendimientos = precios.pct_change().dropna() * 100

print("\nPrimeros cinco rendimientos diarios (%):")
display(rendimientos.head().round(4))

# HISTOGRAMAS
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

for ax, ticker in zip(axes, tickers):
    sns.histplot(
        data=rendimientos,
        x=ticker,
        bins=40,
        kde=True,
        ax=ax
    )

    ax.set_title(f"Distribucion de rendimientos de {ticker}")
    ax.set_xlabel("Rendimiento diario (%)")
    ax.set_ylabel("Frecuencia")

plt.suptitle("Histogramas de rendimientos diarios", fontsize=15)
plt.tight_layout()
plt.show()

# BOXPLOT COMPARATIVO
plt.figure(figsize=(10, 6))

sns.boxplot(data=rendimientos[tickers])

plt.title("Boxplot de rendimientos diarios")
plt.xlabel("Activo")
plt.ylabel("Rendimiento diario (%)")
plt.tight_layout()
plt.show()

# Estadisticas basicas de los rendimientos
print("\nEstadisticas de rendimientos diarios (%):")

estadisticas_rendimientos = pd.DataFrame({
    "Media": rendimientos.mean(),
    "Mediana": rendimientos.median(),
    "Desviacion estandar": rendimientos.std(),
    "Minimo": rendimientos.min(),
    "Maximo": rendimientos.max()
})

display(estadisticas_rendimientos.round(4))

# Identificar activos con mayor y menor volatilidad diaria
volatilidad_diaria = rendimientos.std()

activo_mas_volatil = volatilidad_diaria.idxmax()
activo_menos_volatil = volatilidad_diaria.idxmin()

print(
    f"\nEl activo con mayor volatilidad diaria fue "
    f"{activo_mas_volatil}, con "
    f"{volatilidad_diaria[activo_mas_volatil]:.4f}%."
)

print(
    f"El activo con menor volatilidad diaria fue "
    f"{activo_menos_volatil}, con "
    f"{volatilidad_diaria[activo_menos_volatil]:.4f}%."
)

print(
    "\nInterpretacion: una mayor desviacion estandar indica "
    "mayor variabilidad de los rendimientos diarios. "
    "El boxplot permite comparar la dispersion y detectar "
    "observaciones extremas."
)


# PUNTO 4. MATRIZ DE CORRELACION Y HEATMAP

print("\n" + "=" * 60)
print("PUNTO 4: MATRIZ DE CORRELACION")
print("=" * 60)

# Calcular correlaciones entre los rendimientos diarios
correlacion = rendimientos.corr()

print("\nMatriz de correlacion:")
display(correlacion.round(3))

# Crear heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlacion,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0,
    fmt=".2f",
    square=True,
    linewidths=0.5
)

plt.title("Heatmap de correlacion de los rendimientos")
plt.tight_layout()
plt.show()

# Interpretacion de las correlaciones entre pares
print("\nInterpretacion de las correlaciones:")

for i in range(len(tickers)):
    for j in range(i + 1, len(tickers)):
        activo1 = tickers[i]
        activo2 = tickers[j]
        valor = correlacion.loc[activo1, activo2]

        if valor >= 0.7:
            interpretacion = "relacion positiva fuerte"
        elif valor >= 0.3:
            interpretacion = "relacion positiva moderada"
        elif valor > -0.3:
            interpretacion = "relacion lineal debil"
        elif valor > -0.7:
            interpretacion = "relacion negativa moderada"
        else:
            interpretacion = "relacion negativa fuerte"

        print(
            f"{activo1} y {activo2}: correlacion de "
            f"{valor:.3f}, {interpretacion}."
        )

print(
    "\nInterpretacion financiera: las correlaciones bajas "
    "o negativas pueden ofrecer beneficios de diversificacion. "
    "Una correlacion alta y positiva indica que dos activos "
    "tienden a moverse en la misma direccion."
)

# PUNTO 5. ESTADISTICAS DESCRIPTIVAS Y CONCLUSION

print("\n" + "=" * 60)
print("PUNTO 5: ESTADISTICAS DESCRIPTIVAS")
print("=" * 60)

# Calcular estadisticas de los rendimientos diarios
media_diaria = rendimientos.mean()
mediana_diaria = rendimientos.median()
volatilidad_diaria = rendimientos.std()
mejor_dia = rendimientos.max()
peor_dia = rendimientos.min()

# Anualizar la volatilidad usando 252 dias bursatiles
volatilidad_anualizada = volatilidad_diaria * np.sqrt(252)

# Construir tabla resumen
resumen = pd.DataFrame({
    "Media diaria (%)": media_diaria,
    "Mediana diaria (%)": mediana_diaria,
    "Volatilidad diaria (%)": volatilidad_diaria,
    "Volatilidad anualizada (%)": volatilidad_anualizada,
    "Mejor dia (%)": mejor_dia,
    "Peor dia (%)": peor_dia,
    "Rentabilidad acumulada (%)": rentabilidad_acumulada
})

print("\nTabla de estadisticas descriptivas:")
display(resumen.round(3))

# Comparar rentabilidad y volatilidad
mejor_activo = rentabilidad_acumulada.idxmax()
peor_activo = rentabilidad_acumulada.idxmin()
activo_mas_volatil = volatilidad_diaria.idxmax()
activo_menos_volatil = volatilidad_diaria.idxmin()

print("\n" + "=" * 60)
print("CONCLUSION PARA EL INVERSIONISTA")
print("=" * 60)

print(
    f"1. Mayor rentabilidad acumulada: {mejor_activo} "
    f"({rentabilidad_acumulada[mejor_activo]:.2f}%)."
)

print(
    f"2. Menor rentabilidad acumulada: {peor_activo} "
    f"({rentabilidad_acumulada[peor_activo]:.2f}%)."
)

print(
    f"3. Mayor volatilidad diaria: {activo_mas_volatil} "
    f"({volatilidad_diaria[activo_mas_volatil]:.3f}%)."
)

print(
    f"4. Menor volatilidad diaria: {activo_menos_volatil} "
    f"({volatilidad_diaria[activo_menos_volatil]:.3f}%)."
)

print(
    "\nRecomendacion: el inversionista debe comparar la "
    "rentabilidad acumulada con la volatilidad de cada activo "
    "y analizar la matriz de correlacion para evaluar los "
    "beneficios de diversificacion. La eleccion final debe "
    "ajustarse a su tolerancia al riesgo y a sus objetivos "
    "financieros. Los resultados historicos no garantizan "
    "rendimientos futuros."
)





```
